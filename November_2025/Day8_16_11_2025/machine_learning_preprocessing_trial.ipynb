Using Slicing, locator and ilocator to fetch required rows and columns. Fetching columns, datatypes and other required information of columns

```python
import pandas as pd
df = pd.read_csv("oscar_age_male.csv")
# print(df)
print(df[2:5])#To Slice few rows and return
print(df.iloc[2:5, 3])#To slice few rows and a column using column number to return
print(df.loc[2:5, "Name"])#To slice few rows and a column using column name to return
print(df.columns)#To get column names
print(df.dtypes)#To get the datatypes of each column
print(df.describe())#To get description of dataframe like count, mean, etc.
print(df.info())#information on memory usage, count, datatype, etc.

```

Getting the count of repeated rows and unique rows details

```python
print(df["Name"].value_counts())#gets count of the data repeating
print(df["Name"].unique())#gets name of unique winners
print(df["Name"].nunique())#gives total unique winners
```
