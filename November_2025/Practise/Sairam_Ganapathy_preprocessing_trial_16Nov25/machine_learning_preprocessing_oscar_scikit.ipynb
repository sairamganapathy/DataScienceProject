With SimpleImputer transforming column 'Age'

```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder
pd = df = pd.read_csv("oscar_age_male.csv")
simple_imputer = SimpleImputer(strategy="most_frequent")
df["Age"] = simple_imputer.fit_transform(df[["Age"]])
# print(df.columns)
# print(df[6:9])
```

To set the null value in 'Age Group' column to a categorical value(for setting up my data to have a proper categorical column)

```python
df["Age Group"] = df["Age Group"].astype(str)
df.loc[(df["Age"]<35) & (df["Age Group"].isna()), "Age Group"] = "Youngsters"
df.loc[(df["Age"] > 35) & (df["Age"] <= 60) & (df["Age Group"].isna()), "Age Group"] = "Old Adults"
df.loc[(df["Age"]>60) & (df["Age Group"].isna()), "Age Group"] = "Senior Adults"
print(df)
```

```python
ordinal_encoder = OrdinalEncoder()
df["Age Group"] = ordinal_encoder.fit_transform(df[["Age Group"]])
print(df)
```
